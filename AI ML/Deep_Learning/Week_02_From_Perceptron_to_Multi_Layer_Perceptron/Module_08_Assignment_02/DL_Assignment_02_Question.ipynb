# Question 01: [ Marks 10 ]
Explain gradient descent in detail. What is the main idea behind it? How does it help in minimizing the loss function? Explain the role of learning rate and describe what happens if the learning rate is too high or too low.

**Write** Answer 01:

Gradient descent is an algorithm used to reduce the loss by updating the model's parameters step by step. Its main idea is to move in the opposite direction of the gradient to find the minimum loss. This helps the model improve its predictions.

The learning rate controls how much the parameters are updated. If it is too high, the model may miss the minimum loss. If it is too low, learning becomes very slow.


# Question 02: [ Marks 10 ]

What is forward propagation in an artificial neural network? Explain step by step how data moves from the input layer to the output layer.

## Write Answer 02:

Forward propagation is the process of passing input data through a neural network to get the final output. First, the input data enters the input layer. Then, each hidden layer calculates weighted sums, adds bias, and applies an activation function. Finally, the output layer produces the prediction.


# Question 03: [ Marks 15 ]
Why do we need activation functions in neural networks? What would happen if we only used linear transformations without any activation function?

## Write Answer 03:
Activation functions help a neural network learn complex patterns. They add non-linearity, so the model can solve difficult problems.

If we only use linear transformations, the whole network behaves like a single linear model, no matter how many layers it has. As a result, it cannot learn complex relationships and its performance becomes limited.

# Question 04: [ Marks 15 ]

Explain the sigmoid activation function. Where is it used? What are its advantages and disadvantages? What problem does it cause during training?

## Write Answer 04:

The sigmoid activation function converts input values into a range between 0 and 1. It is mainly used in the output layer for binary classification.

Its advantage is that it gives probability-like outputs. Its disadvantage is that it is slow and can cause the **vanishing gradient** problem. During training, this problem makes gradients very small, so the model learns slowly.


# Question 05: [ Marks 10 ]
Explain the softmax activation function in simple words. Why is it used in multi-class classification problems? How is it different from sigmoid?

## Write Answer 05:
The softmax activation function converts the output values into probabilities, where the total is always 1. It is used in multi-class classification to choose the most likely class.

Unlike sigmoid, which is used for binary classification and gives one probability, softmax compares all classes and gives a probability for each class.

# Question 02: [ Marks 20 ]

Manual Gradient Calculation

Given:

z = w*x + b <br>
y_pred = sigmoid(z) <br>
L = −[ y log(y_pred) + (1−y) log(1−y_pred) ] <br>

a) Write dL/dy_pred <br>
b) Write dy_pred/dz <br>
c) Show that dL/dz = y_pred − y <br>

## Write Answer 02:

a) dL/dy_pred

L = -[y·log(y_pred)+(1−y)·log(1−y_pred)]

dL/dy_pred = -[ y/y_pred − (1−y)/(1−y_pred)]

dL/dy_pred = (y_pred - y) / [y_pred(1−y_pred)]

b) dy_pred/dz

y_pred = sigmoid(z) = 1/(1+e^(-z))

dy_pred/dz = sigmoid(z)·(1-sigmoid(z)) = y_pred(1-y_pred)

c) dL/dz = y_pred-y
Using chain rule:

dL/dz = dL/dy_pred * dy_pred/dz

dL/dz = [(y_pred−y) / (y_pred(1−y_pred))] * [y_pred(1−y_pred)]

The y_pred(1-y_pred) term in numerator and denominator cancels out:

dL/dz = y_pred-y
(Proved)

c) Explain how would you do it using Autograd

Autograd (used in PyTorch/TensorFlow) calculates gradients automatically, so we don't need to do the derivative by hand. It works like this:

1. During the forward pass, every operation (x², 4x, addition) gets recorded in a computation graph.
2. When we call `.backward()`, Autograd goes through this graph in reverse order and automatically applies the chain rule to find the gradient at each node.
3. Finally `x.grad` gives us the value of dL/dx, which matches the manual result (10 in this case).

# Question 04: [ Marks 15 ]

```python
# Answer 04:
import torch
x = torch.tensor(3.0, requires_grad=True)
y = x**2 + 4*x

y.backward()
print("Gradient of X:", x.grad)
```

Create two tensors:

b = 1.0 (requires_grad=True) <br>

Let input x = 3.0. <br>
Compute z = w*x + b. <br>
Use Sigmoid to get y_pred.
 <br>

Define Binary Cross Entropy manually (without built-in loss). <br>
Compute gradients using backward(). <br>
Print gradients of w and b. <br>

Write full code.

## Write Answer 05:

```python
import torch
w = torch.tensor(2.0, requires_grad=True)
b = torch.tensor(1.0, requires_grad=True)

x = torch.tensor(3.0)
z = w*x+b
y_pred = torch.sigmoid(z)
y = torch.tensor(1.0)
loss = -(y * torch.log(y_pred) + (1-y) * torch.log(1-y_pred))

loss.backward()
print("Gradient of w:", w.grad)
print("Gradient of b:", b.grad)
```

# Question 07: [ Marks 20 ]


Create tensor w = 2.0 (requires_grad=True). <br>

Compute two different losses sequentially and call backward() twice. <br>
Print w.grad. <br>

Do not clear gradients. <br>
Explain what you observe. <br>

Write full code.


## Write Answer 07:



```python
import torch
w = torch.tensor(2.0, requires_grad=True)
L1 = w**2
L1.backward()
print("Gradient W after L1.backward():", w.grad)

L2 = 3*w
L2.backward()
print("Gradient W after L2.backward()", w.grad)
```

Explain what you observe:

After the second `backward()`, `w.grad` becomes 7, not 3.

This is because PyTorch accumulates gradients by default instead of replacing them. Since we didn't clear gradients with `zero_grad()`, the new gradient (3) got added to the old one (4), giving 4 + 3 = 7.

This is why we must call `zero_grad()` before each `backward()` in training, otherwise gradients keep adding up wrongly.

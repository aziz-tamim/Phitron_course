# Question 01: [ Marks 10 ]

## Explain the concept of backpropagation in detail. Describe how the error is propagated from the output layer to the hidden layers. Derive the mathematical expressions for weight updates using the chain rule, and explain how learning rate affects the training process.


# Write Answer 01:

Backpropagation is the method used to train a neural network by reducing the error between predicted and actual output.

It works in two steps. In the forward pass, input passes through the network and produces an output ŷ. This is compared with the actual value y to calculate the loss:

L = (1/2) (y - ŷ)²

In the backward pass, this error is sent back from the output layer toward the hidden layers, to find out how much each weight caused the error. This is done using the chain rule:

∂L/∂w = (∂L / ∂ŷ) * (∂ŷ / ∂z) * (∂z/∂w)

= (ŷ - y) * ŷ(1 - ŷ) * x

For hidden layer weights, the same chain rule applies, but the gradient also depends on the error already calculated in the layer ahead of it. That's why the error keeps flowing backward, layer by layer hence the name backpropagation.

Once gradients are found, weights are updated using Gradient Descent:

w_new = w_old - η(∂L/∂w)

Here η (eta) is the learning rate, which decides how big a step is taken during the update.

Too high -> may overshoot and never converge

Too low -> training becomes very slow

Balanced -> smooth and steady convergence to minimum loss

δ1 = (δ2 * W2) * a1(1 - a1)

∂L/∂W1 = δ1 * x

∂L/∂b1 = δ1

Weight Update (Gradient Descent):

W2_new = W2_old - η(∂L/∂W2)

b2_new = b2_old - η(∂L/∂b2)

W1_new = W1_old - η(∂L/∂W1)

b1_new = b1_old - η(∂L/∂b1)

The error is first calculated at the output layer (δ2), then passed backward to the hidden layer (δ1) using the output weights W2. Each layer's gradient depends on its local derivative and the error coming from the layer ahead that's why it's called backpropagation.


# Question 03: [Marks 5]

## Discuss the vanishing gradient problems in deep neural networks. Explain why they occur during backpropagation, how activation functions contribute to this issue, and suggest at least three techniques to overcome these problems.


## Write Answer 03:

**Vanishing Gradient Problem**

In deep neural networks, backpropagation uses the chain rule, multiplying many derivatives together as the error moves backward from the output layer to earlier layers. When these derivatives are small (less than 1), multiplying many of them makes the gradient shrink more and more. By the time it reaches the earlier layers, the gradient becomes almost zero. This is the vanishing gradient problem. Because of this, earlier layers barely get updated, and training becomes very slow or stops improving.

**Why activation functions cause this:**

Activation functions like Sigmoid and Tanh squeeze output into a small range, and their derivatives are also small sigmoid's maximum derivative is only 0.25. When many layers use these, the small derivatives keep multiplying during backpropagation, making the gradient shrink rapidly.

**Techniques to overcome it:**

Use ReLU activation instead of sigmoid/tanh its derivative is 0 or 1, so it doesn't shrink the gradient much.
Use proper weight initialization (Xavier or He initialization) keeps values in a good range from the start.
Use Batch Normalization normalizes each layer's input, helping gradients flow better.

## Describe the architecture of an Artificial Neural Network (ANN) in detail. Explain the role of each component (neurons, weights, bias, activation functions). Also discuss how increasing the number of hidden layers and neurons affects model performance and overfitting.


## Write Answer 04:

**Artificaial Neural Network(ANN)**

An ANN is made up of three types of layers connected together:

Input Layer receives the raw data (features) and passes it forward. No computation happens here.

Hidden Layer(s) process the data using weights, biases, and activation functions to find patterns.

Output Layer gives the final prediction or result.

Each neuron in one layer connects to neurons in the next layer, and every connection has a weight attached to it.

**Role of Each Component**

Neurons: The basic units that receive input, apply weights and bias, then pass the result through an activation function to produce output.

Weights: Control how much influence each input has on the output. Higher weight = more importance. Weights are updated during training to reduce error.

Bias: An extra adjustable value added to the weighted sum. It lets the model shift the activation function, giving more flexibility to fit the data properly.

Activation Function: Adds non-linearity to the network (e.g.:- ReLU, Sigmoid, Tanh). Without it, the whole network would behave like a simple linear model, even with many layers.

**Effect of Increasing Hidden Layers and Neurons**

Adding more hidden layers and neurons increases the network's capacity to learn complex patterns, which can improve performance on training data.

But if there are too many layers/neurons compared to the amount of data available, the model starts overfitting it memorizes the training data instead of learning general patterns. This makes it perform well on training data but poorly on new, unseen data.

So the number of layers and neurons should be chosen based on the complexity of the problem and dataset size, using techniques like dropout or regularization to keep overfitting under control.

## Compare Artificial Neural Networks (ANN) and Convolutional Neural Networks (CNN) in detail. Discuss their architectures, working principles, advantages, limitations, and typical use cases such as image processing, tabular data, and pattern recognition.

**Architecture**

ANN: Fully connected layers — input, hidden, and output layers. Every neuron connects to every neuron in the next layer.

CNN: Uses convolutional layers (extract features using filters), pooling layers (reduce size), and fully connected layers at the end for classification.

**Working Principle**

ANN: Takes input as a flat vector of numbers and processes it layer by layer using weights and activation functions.

CNN: Applies filters that slide over the input (usually images) to detect local patterns like edges and shapes, then combines them in deeper layers to recognize complex features.

**Advantages:**

ANN: Simple, easy to design, works well on structured/tabular data.

CNN: Automatically learns spatial features, preserves spatial relationships between pixels, uses fewer parameters through weight sharing.

**Limitations**

ANN: Loses spatial structure on images, needs huge parameters for large inputs, causing overfitting.

CNN: Needs more computation power and larger datasets, more complex to design.

**Use Cases**

ANN: Tabular data (house price prediction, customer churn), simple classification problems.

CNN: Image processing tasks image classification, object detection, facial recognition.

# Question 06: [ Marks 10 ]

## Explain different activation functions used in neural networks, including Sigmoid, Tanh, and ReLU. Provide their mathematical formulas, advantages, disadvantages, and suitable use cases. Also discuss why ReLU is widely used in deep learning.

## Write Answer 06:

1. Sigmoid

Formula: σ(x) = 1 / (1 + e^(-x))

Range: 0 to 1
Advantages: Smooth output, useful for probability-based outputs (binary classification).
Disadvantages: Suffers from vanishing gradient problem, output is not zero-centered, computationally expensive (uses exponential).
Use case: Output layer of binary classification models.

2. Tanh

Formula: tanh(x) = (e^x - e^(-x)) / (e^x + e^(-x))

Range: -1 to 1
Advantages: Zero-centered output (better than sigmoid), stronger gradients than sigmoid.
Disadvantages: Still suffers from vanishing gradient problem for large positive/negative inputs.
Use case: Hidden layers, especially in RNNs.

3. ReLU (Rectified Linear Unit)

Formula: f(x) = max(0, x)

Range: 0 to ∞
Advantages: Computationally simple and fast, doesn't saturate for positive values, reduces vanishing gradient problem.
Disadvantages: "Dying ReLU" problem neurons can get stuck outputting 0 if inputs are always negative.
Use case: Hidden layers of most deep learning models (CNNs, ANNs).

Why ReLU is widely used

ReLU is preferred because it's simple and fast to compute (no exponential calculation like sigmoid/tanh). It doesn't squeeze large input values into a small range, so gradients stay strong during backpropagation, which helps avoid the vanishing gradient problem. This allows deep networks to train faster and more effectively compared to using sigmoid or tanh.

Convolution Operation in CNNs

Convolution is the core operation that lets a CNN extract features (like edges, corners, textures) from an image. Instead of connecting every pixel to every neuron (like ANN), a small matrix called a filter (kernel) slides over the input and performs element-wise multiplication with the part of the input it's currently covering, then sums the result into a single number. This is repeated as the filter slides across the whole image, producing a feature map.

How filters work:

A filter is a small matrix of weights. As it slides over the input, it detects specific patterns one filter might detect vertical edges, another might detect horizontal edges, depending on what values it learns during training.

Step-by-step example (shown in the image above):

Input: 5x5 matrix
Filter: 3x3 matrix
Stride: 1 (filter moves 1 step at a time)
Padding: none

At each position, the filter overlaps a 3x3 region of the input. We multiply each overlapping pair of values and sum them this single number becomes one cell of the output feature map. For the first (top-left) position: multiplying and summing all 9 overlapping values gives 4, as shown in the image. The filter then slides right by 1 (stride 1) and repeats, continuing until it has covered the whole input, producing a 3x3 output feature map.

Effect of Stride and Padding:

Stride: the number of pixels the filter moves each step. A stride of 1 moves slowly and produces a larger output (more detail). A larger stride skips more pixels, producing a smaller output and reducing computation.
Padding: adding extra pixels (usually zeros) around the input border. Without padding ("valid" padding), the output shrinks after each convolution. With padding ("same" padding), zeros are added so the output size stays the same as the input.

Output size formula:

Output size = ((Input size - Filter size + 2*Padding) / Stride) + 1

For our example: (5 - 3 + 0)/1 + 1 = 3, matching the 3x3 output shown.

## Explain the concept of pooling in CNNs. Compare MaxPooling and AveragePooling in terms of operation, advantages, and impact on feature extraction. Discuss how pooling helps in reducing overfitting and computational complexity.

## Write Answer 08:

Pooling in CNNs

Pooling is a downsampling operation applied after convolution layers. It reduces the spatial size (width and height) of the feature map while keeping the most important information, making the network faster and less prone to overfitting.

Pooling works by sliding a small window over the feature map and summarizing the values inside that window into a single number, similar to how a filter slides in convolution, but with no learnable weights.

MaxPooling vs AveragePooling

MaxPooling

Operation: Takes the maximum value from each window.

Advantage: Keeps the strongest/most prominent feature (like a sharp edge), making it good at capturing dominant patterns. Works well in practice for most image tasks.

Impact on feature extraction: Preserves the most activated (important) features, ignores weaker ones helps retain key patterns even when there's small distortion or shift in the image.

AveragePooling

Operation: Takes the average of all values in each window.

Advantage: Gives a smoother, more generalized representation, retains overall context of the region instead of just the peak value.

Impact on feature extraction: Can dilute strong features since it blends them with weaker ones, so it sometimes loses sharp/important details that MaxPooling would keep.

In practice, MaxPooling is more commonly used because it usually gives better performance for image classification tasks by keeping the most distinctive features.

How Pooling Reduces Overfitting and Computational Complexity

Reduces overfitting: By shrinking the feature map, pooling reduces the number of parameters in later layers, so the model has less chance of memorizing noise/details specific to the training data, and generalizes better to new data. It also makes the network slightly more tolerant to small shifts/distortions in the input (translation invariance).

Reduces computational complexity: Smaller feature maps mean fewer computations and less memory needed in the following layers, making training and inference faster.

# Question 10: [ Marks 20 ]

## Design and implement a deep Convolutional Neural Network (CNN) for a multi-class image classification problem using PyTorch. The network should consist of four convolutional blocks, where each block includes two convolution layers followed by activation and normalization, and a pooling layer at the end. The number of filters should increase progressively across blocks. After the convolutional part, flatten the feature maps and connect them to fully connected layers with dropout for regularization, and produce final outputs corresponding to K classes. Also explain the role of each part of the network in detail.

# Write Answer 10:

```python
import torch
import torch.nn as nn

class DeepCNN(nn.Module):
  def __init__(self, num_classes):
    super(DeepCNN, self).__init__()

    self.block1 = nn.Sequential(
        nn.Conv2d(3, 32, kernel_size=3, padding=1),
        nn.BatchNorm2d(32),
        nn.ReLU(),
        nn.Conv2d(32, 32, kernel_size=3, padding=1),
        nn.BatchNorm2d(32),
        nn.ReLU(),
        nn.MaxPool2d(kernel_size=2, stride = 2)
    )

    self.block2 = nn.Sequential(
        nn.Conv2d(32, 64, kernel_size=3, padding=1),
        nn.BatchNorm2d(64),
        nn.ReLU(),
        nn.Conv2d(64, 64, kernel_size=3,padding=1),
        nn.BatchNorm2d(64),
        nn.ReLU(),
        nn.MaxPool2d(kernel_size=2, stride=2)
    )
    self.block3 = nn.Sequential(
        nn.Conv2d(64, 128, kernel_size=3, padding=1),
        nn.BatchNorm2d(128),
        nn.ReLU(),
        nn.Conv2d(128, 128, kernel_size=3, padding=1),
        nn.BatchNorm2d(128),
        nn.ReLU(),
        nn.MaxPool2d(kernel_size=2, stride = 2)
    )

    self.block4 = nn.Sequential(
        nn.Conv2d(128, 256, kernel_size=3, padding=1),
        nn.BatchNorm2d(256),
        nn.ReLU(),
        nn.Conv2d(256, 256, kernel_size = 3, padding=1),
        nn.BatchNorm2d(256),
        nn.ReLU(),
        nn.MaxPool2d(kernel_size=2, stride=2)
    )

    self.fc = nn.Sequential(
        nn.Flatten(),
        nn.Linear(256* 8 * 8, 512),
        nn.ReLU(),
        nn.Dropout(0.5),
        nn.Linear(512, num_classes)
    )

  def forward(self, x):
    x = self.block1(x)
    x = self.block2(x)
    x = self.block3(x)
    x = self.block4(x)
    x = self.fc(x)
    return x

model = DeepCNN(num_classes = 10)
print(model)
```

Role of Each Part

Convolution layers (Conv2d) Extract features from the image like edges, textures, and shapes. Two conv layers per block let the network learn richer features before downsampling.

Batch Normalization (BatchNorm2d) Normalizes the output of each conv layer, stabilizing and speeding up training, and helping avoid vanishing/exploding gradients.

Activation (ReLU) Adds non-linearity, allowing the network to learn complex, non-linear patterns.

Pooling (MaxPool2d) Placed at the end of each block, reduces spatial size (height, width), keeping the strongest features and reducing computation.

Increasing filters (32 -> 64 -> 128 -> 256) Early layers detect simple patterns (edges, corners) with fewer filters. Deeper layers detect more complex, abstract features, so more filters are used.

Flatten Converts the 3D feature map (channels * height * width) into a 1D vector to feed into the fully connected layers.

Fully Connected Layers (Linear) Combine all extracted features to make the final classification decision.

Dropout Randomly disables some neurons during training to reduce overfitting.

Output Layer Final Linear layer outputs K values (one score per class), later converted to probabilities using Softmax (usually inside CrossEntropyLoss).

### Question 01:

**1: Forget Gate এর উদ্দেশ্য**

`ft = σ(Wf · [ht-1, xt] + bf)`

Explanation

Forget Gate ঠিক করে দেয় — আগের cell state (Ct-1) থেকে কতটুকু তথ্য রাখা হবে আর কতটুকু ভুলে যাওয়া হবে। এটা আগের hidden state (ht-1) এবং বর্তমান input (xt) কে একসাথে নিয়ে sigmoid ফাংশনের মধ্য দিয়ে পাঠায়, যার আউটপুট থাকে 0 থেকে 1 এর মধ্যে।

0 → সেই তথ্যটা সম্পূর্ণ ভুলে যাও

1 → সেই তথ্যটা সম্পূর্ণ রেখে দাও

এভাবে LSTM অপ্রয়োজনীয় পুরনো তথ্য মুছে ফেলে মেমোরিকে পরিষ্কার রাখে।

**2. Input Gate এর ভূমিকা**

Formulas:

`it = σ(Wi · [ht-1, xt] + bi)`

`C̃t = tanh(Wc · [ht-1, xt] + bc)`

Explanation:
Input Gate ঠিক করে নতুন কোন তথ্য cell state এ যোগ করা হবে।

it (sigmoid দিয়ে গণনা) → ঠিক করে কতটুকু নতুন তথ্য গুরুত্বপূর্ণ, মান থাকে 0 থেকে 1 এর মধ্যে (একটা filter হিসেবে কাজ করে)
C̃t (tanh দিয়ে গণনা) → নতুন candidate তথ্য তৈরি করে, মান থাকে -1 থেকে 1 এর মধ্যে (সম্ভাব্য নতুন মেমোরি)

এই দুটো মিলে ঠিক হয় — কোন নতুন তথ্য কতটুকু গুরুত্ব দিয়ে মেমোরিতে যোগ হবে।

**3. Cell State আপডেট**

Formula:

`Ct = ft * Ct-1 + it * C̃t`

Explanation:-

এটাই LSTM এর "memory update" ধাপ। দুইটা অংশ:

ft * Ct-1 → আগের memory থেকে যতটুকু রাখতে চাই (Forget Gate অনুযায়ী)

it * C̃t → নতুন যতটুকু তথ্য যোগ করতে চাই (Input Gate অনুযায়ী)

এই দুইটার যোগফলই হলো নতুন cell state Ct। এভাবে পুরনো দরকারি তথ্য আর নতুন গুরুত্বপূর্ণ তথ্য একসাথে মিলে LSTM এর দীর্ঘমেয়াদি মেমোরি তৈরি হয়।

**4. Output Gate এর উদ্দেশ্য**

Formulas:

`ot = σ(Wo · [ht-1, xt] + bo)`

`ht = ot * tanh(Ct)`

Explanation:

Output Gate ঠিক করে cell state Ct থেকে কতটুকু তথ্য বর্তমান hidden state (ht) হিসেবে বাইরে পাঠানো হবে।

ot → sigmoid দিয়ে ঠিক হয় কোন অংশ output হিসেবে যাবে

tanh(Ct) → cell state কে -1 থেকে 1 এর মধ্যে normalize করে
দুটোর গুণফল (ht) হলো এই timestep এর চূড়ান্ত output, যেটা পরের timestep এ এবং প্রয়োজনে prediction এ ব্যবহার হয়।

**5. তিনটা Gate কেন দরকার?**

LSTM এর মূল লক্ষ্য হলো long-term dependency মনে রাখা, আর এই কাজটা করতে তিনটা আলাদা দায়িত্ব দরকার হয়:

Forget Gate — পুরনো অপ্রয়োজনীয় তথ্য মুছে ফেলা
`ft = σ(Wf[ht-1, xt] + bf)`

Input Gate — নতুন গুরুত্বপূর্ণ তথ্য যোগ করা
`it = σ(Wi[ht-1, xt] + bi)`, `C̃t = tanh(Wc[ht-1, xt] + bc)`

Output Gate — বর্তমান output/hidden state নির্ধারণ করা
`ot = σ(Wo[ht-1, xt] + bo)`

একটা মাত্র gate দিয়ে এই তিনটা কাজ (মুছে ফেলা, যোগ করা, আউটপুট দেওয়া) একসাথে ভালোভাবে করা যায় না। তাই আলাদা আলাদা gate ব্যবহার করে LSTM কে নিয়ন্ত্রণযোগ্য এবং স্থিতিশীল memory ব্যবস্থাপনা দেওয়া হয়েছে। এভাবেই এটা Vanilla RNN এর vanishing gradient সমস্যা সমাধান করে।

**6. তিনটা Gate এর মধ্যে পার্থক্য**

Forget Gate

+ কাজ: পুরনো তথ্য রাখবে না ভুলবে ঠিক করে
+ Formula: ft = σ(Wf[ht-1, xt] + bf)
+ কাজ করে: Ct-1 (আগের cell state) এর উপর
+ Activation: Sigmoid

Input Gate

+ কাজ: নতুন তথ্য কতটুকু যোগ হবে ঠিক করে
+ Formula: it = σ(Wi[ht-1, xt] + bi), C̃t = tanh(Wc[ht-1, xt] + bc)
+ কাজ করে: নতুন candidate memory এর উপর
+ Activation: Sigmoid + Tanh

Output Gate

+ কাজ: hidden state হিসেবে কতটুকু বের হবে ঠিক করে
+ Formula: ot = σ(Wo[ht-1, xt] + bo)
+ কাজ করে: বর্তমান Ct এর উপর
+ Activation: Sigmoid (+ Tanh চূড়ান্ত output এ)

**7. কীভাবে Forget Gate পুরনো এবং Input Gate নতুন মেমোরি নিয়ন্ত্রণ করে**

পুরনো মেমোরি (Forget Gate দিয়ে নিয়ন্ত্রণ):
`ft * Ct-1`

এখানে ft এর মান 0 এর কাছাকাছি হলে আগের memory অনেকটাই মুছে যাবে, আর 1 এর কাছাকাছি হলে আগের memory প্রায় পুরোটাই থেকে যাবে।

নতুন মেমোরি (Input Gate দিয়ে নিয়ন্ত্রণ):
`it * C̃t`

এখানে C̃t হলো candidate নতুন তথ্য, আর it ঠিক করে সেই candidate তথ্যের কতটুকু আসলে cell state এ যোগ হবে।

এই দুটো মিলিয়েই চূড়ান্ত cell state তৈরি হয়:
`Ct = ft * Ct-1 + it * C̃t`

অর্থাৎ, একদিকে পুরনো তথ্য filter হচ্ছে, আরেকদিকে নতুন তথ্য যোগ হচ্ছে — দুটো প্রক্রিয়া সমান্তরালে চলে।

**8. Output Gate কীভাবে Hidden State নির্ধারণ করে**

`ot = σ(Wo[ht-1, xt] + bo)`
`ht = ot * tanh(Ct)`

প্রথমে updated cell state Ct কে tanh দিয়ে -1 থেকে 1 এর মধ্যে scale করা হয় — এতে এটা একটা normalized representation পায়। তারপর ot (যেটা sigmoid দিয়ে বলে দেয় কোন অংশ গুরুত্বপূর্ণ) দিয়ে গুণ করা হয়। ফলে ht হলো cell state এর একটা selective/filtered version, যেটা শুধু প্রাসঙ্গিক তথ্যটুকুই বহন করে — পরবর্তী timestep এবং output prediction এর জন্য।

**9. তিনটা লাইনের ব্যাখ্যা formula সহ**

"Forget Gate removes unnecessary information, Input Gate adds important information, and Output Gate controls the final output."

Forget Gate অপ্রয়োজনীয় তথ্য সরায়:
`ft = σ(Wf[ht-1, xt] + bf) → ft * Ct-1`
যদি কোনো তথ্য এখন আর দরকার না হয়, ft শূন্যের কাছাকাছি হয়ে সেটাকে মুছে দেয়।

Input Gate গুরুত্বপূর্ণ তথ্য যোগ করে:
`it = σ(Wi[ht-1, xt] + bi)`, `C̃t = tanh(Wc[ht-1, xt] + bc) → it * C̃t`
নতুন input থেকে যেটুকু গুরুত্বপূর্ণ মনে হয়, সেটুকু cell state এ যোগ হয়।

Output Gate চূড়ান্ত output নিয়ন্ত্রণ করে:
`ot = σ(Wo[ht-1, xt] + bo)` → `ht = ot * tanh(Ct)`
আপডেট হওয়া cell state থেকে কতটুকু বাইরে পাঠানো হবে (hidden state হিসেবে), সেটা এই gate ঠিক করে।

তিনটা মিলে একটা সুশৃঙ্খল মেমোরি ব্যবস্থাপনা তৈরি করে — মুছে ফেলা, যোগ করা, এবং প্রকাশ করা।

**10. LSTM Cell এর সম্পূর্ণ Workflow**

একটা timestep t এ LSTM এর ভিতরে ধাপে ধাপে যা ঘটে:

ধাপ ১ — Forget Gate: আগের cell state থেকে কী রাখা হবে ঠিক করা
`ft = σ(Wf[ht-1, xt] + bf)`

ধাপ ২ — Input Gate: নতুন কী তথ্য যোগ হবে ঠিক করা
`it = σ(Wi[ht-1, xt] + bi)`
`C̃t = tanh(Wc[ht-1, xt] + bc)`

ধাপ ৩ — Cell State Update: পুরনো + নতুন মিলিয়ে memory আপডেট করা
`Ct = ft * Ct-1 + it * C̃t`

ধাপ ৪ — Output Gate: আপডেট হওয়া memory থেকে output/hidden state বের করা
`ot = σ(Wo[ht-1, xt] + bo)`
`ht = ot * tanh(Ct)`

সংক্ষেপে flow:
`(ht-1, xt) → ft, it, C̃t, ot → Ct → ht`

এই ht এবং Ct ই পরের timestep এ ইনপুট হিসেবে যায়, আর এভাবেই LSTM ধারাবাহিকভাবে sequence এর মধ্যে দিয়ে গুরুত্বপূর্ণ তথ্য বহন করে নিয়ে যায় — long-term dependency সমস্যা সমাধান করে।

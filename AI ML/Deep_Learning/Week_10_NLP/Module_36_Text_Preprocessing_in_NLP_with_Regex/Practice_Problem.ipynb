## Data Acquisition

### Data Acquisition কী?

Ans:- Data Acquisition মানে হলো কোনো NLP problem solve করার জন্য প্রয়োজনীয় raw data (text) সংগ্রহ করা। Pipeline-এর প্রথম ধাপ এটা।

### কেন data collect করতে হয়?

Ans:- কারণ machine learning/NLP model শেখে data থেকে (pattern খুঁজে বের করে)। Data ছাড়া model train করা সম্ভব না — "no data, no model"।

### Data কোথা থেকে collect করা যায়?

+ Public datasets (Kaggle, UCI, HuggingFace Datasets)
+ Web scraping (news site, blog, social media)
+ APIs (Twitter API, Reddit API)
+ Company/organization এর নিজস্ব database
+ Surveys বা manual annotation
+ Open-source corpora (Wikipedia dump, Common Crawl)

### একটা dataset কী কারণে "good" হয়?

+ Problem-এর সাথে relevant হতে হবে
+ পর্যাপ্ত পরিমাণে data (enough volume)
+ Diverse/balanced (সব class/category properly represented)
+ Clean এবং কম noise
+ Correctly labeled (supervised task হলে)
+ Bias কম থাকা

## Text Preparation

### Text Preparation কী?

Raw text-কে model-এর জন্য usable format-এ নিয়ে আসার process — এর মধ্যে cleaning এবং preprocessing দুটোই পড়ে।

### Text cleaning কেন জরুরি?

Raw text-এ অনেক noise থাকে (HTML tag, special character, typo, extra space) — এগুলো model-কে confuse করে এবং performance কমিয়ে দেয়। Clean data = better learning।

### Raw text-এ কী কী cleaning step করা হয়?

+ HTML tags/URLs remove করা
+ Lowercasing
+ Extra whitespace/newline remove
+ Special characters/punctuation remove
+ Emoji/non-ASCII character handle করা
+ Duplicate rows drop করা

### Text preprocessing কী?

Cleaning-এর পরের ধাপ — text-কে model-friendly numerical/structured form-এ prepare করার কাজ (tokenization, normalization ইত্যাদি)।

### তুমি এখন পর্যন্ত কী কী preprocessing technique শিখেছ?

+ Tokenization
+ Lowercasing
+ Stemming

### Advanced preprocessing কখন লাগে?

যখন data domain-specific হয় (medical, legal text), অথবা language complex হয় (Bangla-এর মতো morphologically rich language), তখন POS tagging, Named Entity Recognition (NER), spell correction, বা custom tokenizer লাগতে পারে।


## Regular Expressions (Regex)

### Regex কী?

Regex হলো এমন একটি pattern-matching language, যা দিয়ে text-এর মধ্যে specific pattern খুঁজে বের করা, match করা বা replace করা যায়।

### NLP-তে Regex কেন ব্যবহার করি?

Structured pattern (email, phone number, date, URL) দ্রুত detect বা extract করার জন্য, এবং rule-based text cleaning-এর জন্য।

### Regex pattern কী?

একটা string যা define করে দেয় কী ধরনের character sequence আমরা match করতে চাই। যেমন: `\d+` মানে এক বা একাধিক digit।

`.`, `*`, `+`, `?`, `^`, `$` symbol-গুলোর কাজ কী?

+ `.` → যেকোনো একটা character (newline ছাড়া)
+ `*` → আগের character 0 বার বা তার বেশি
+ `+` → আগের character 1 বার বা তার বেশি
+ `?` → আগের character 0 অথবা 1 বার (optional)
+ `^` → string-এর শুরু (start)
+ `$` → string-এর শেষ (end)

### Regex দিয়ে email/phone/URL কীভাবে খুঁজে বের করা যায়?

+ Email: `\w+@\w+\.\w+`
+ Phone: `\d{11}`
+ URL: `https?://\s+`

### Regex text cleaning-এ কীভাবে সাহায্য করে?

Unwanted characters (punctuation, digit, special symbol) খুঁজে বের করে remove/replace করতে regex ব্যবহার হয় — যেমন `re.sub(r'[^a-zA-Z\s]', '', text)`।

## Feature Engineering

### Feature Engineering কী?
Raw text/data থেকে meaningful numerical features তৈরি করার process, যা model বুঝতে ও ব্যবহার করতে পারে।

### Model কেন raw text সরাসরি নিতে পারে না?
Machine learning model শুধু numbers (vectors) নিয়ে কাজ করে — text একটা unstructured string, model সেটা mathematically process করতে পারে না। তাই text-কে numeric representation-এ convert করতে হয় (Bag of Words, TF-IDF, Word2Vec ইত্যাদি)।

### কোন representation technique কখন বেছে নেবে?
+ ছোট dataset, simple task → Bag of Words / TF-IDF
+ Semantic meaning দরকার → Word Embeddings (Word2Vec, GloVe)
+ Context-aware/complex task → Transformer-based embeddings (BERT)
+ Speed priority হলে simple technique, accuracy priority হলে advanced technique।


## Modelling

### Modelling কী?
Prepared data থেকে pattern শেখার জন্য একটা algorithm apply করা, যাতে সেটা prediction/classification করতে পারে।

### Algorithm কীভাবে বেছে নেওয়া হয়?
Problem type (classification/regression), dataset size, data-এর nature (text/image), computational resource, এবং accuracy vs interpretability trade-off দেখে।

### Text classification-এর জন্য কোন algorithm এবং কেন?
+ Simple/small data → Naive Bayes বা Logistic Regression (fast, interpretable)
+ Large data, complex pattern → LSTM বা Transformer (BERT) — কারণ এগুলো context ভালো বোঝে

### Training এবং Testing-এর মধ্যে পার্থক্য কী?
+ Training: model data থেকে pattern শেখে (parameters adjust হয়)
+ Testing: unseen data দিয়ে model-এর performance যাচাই করা হয় — model ওই সময় শেখে না, শুধু predict করে

## Evaluation
### Model evaluation কী?
Model কতটা ভালো কাজ করছে সেটা measure করা, বিভিন্ন metric ব্যবহার করে।

### কেন evaluate করি?
Model-এর strength/weakness বুঝতে, deployment-এর আগে confirm করতে যে model reliable কিনা।

### Accuracy কী?
মোট prediction-এর মধ্যে সঠিক prediction-এর percentage।
`Accuracy = (Correct Predictions) / (Total Predictions)`

## Precision কী?
যতগুলো prediction "positive" বলা হয়েছে, তার মধ্যে আসলে কতগুলো সঠিকভাবে positive।
`Precision = TP / (TP + FP)`

## Recall কী?
আসল positive-গুলোর মধ্যে model কতগুলো সঠিকভাবে ধরতে পেরেছে।
`Recall = TP / (TP + FN)`

## F1-Score কী?
Precision এবং Recall-এর harmonic mean — দুটোর balance দেখায়।
`F1 = 2 * (Precision * Recall) / (Precision + Recall)`

## Accuracy কখন ভালো metric না?
যখন dataset imbalanced হয় (যেমন 95% negative, 5% positive) — তখন সবকিছু "negative" predict করলেও accuracy বেশি দেখাবে, কিন্তু model আসলে useless। তখন Precision, Recall, F1-Score দেখা উচিত।

## Extrinsic Evaluation

### Extrinsic Evaluation কী?
Model-কে বাস্তব application-এর মধ্যে বসিয়ে তার real-world impact/usefulness যাচাই করা (শুধু metric না দেখে)।

### Real-world-এ model useful কিনা কীভাবে বুঝব?
User satisfaction, business outcome (যেমন conversion rate বাড়া), task completion rate, বা end-to-end system performance দেখে।

###  ভালো accuracy থাকা সত্ত্বেও model fail করতে পারে কেন?
হ্যাঁ পারে — কারণ test data হয়তো real-world data-কে ঠিকমতো represent করে না, অথবা edge cases handle করতে পারে না, অথবা user experience খারাপ (slow, confusing output)।

## Deployment
### Model Deployment কী?
Trained model-কে real users বা system-এর জন্য accessible করে তোলা (production environment-এ launch করা)।

### কেন deploy করি?
Model কে actual value দিতে হলে সেটা end-user-দের ব্যবহারের জন্য available করতে হয় — শুধু notebook-এ রাখলে কোনো লাভ নেই।

### User কীভাবে deployed model-এর সাথে interact করে?
+ Web app/UI (যেমন Gradio, Streamlit)
+ REST API call
+ Mobile app integration
+ Chatbot interface

### Deployment-এর পর কী কী challenge হতে পারে?
+ Scalability issue (অনেক user একসাথে ব্যবহার করলে)
+ Latency/slow response
+ Real-world data training data থেকে different হওয়া (data drift)
+ Server downtime/maintenance cost

## Monitoring

### Model Monitoring কী?
Deployment-এর পর model-এর performance continuously track করা, যাতে সময়ের সাথে degrade না করে।

### Monitoring কেন জরুরি?
Real-world data সময়ের সাথে বদলায় (user behavior, trend), model সেটার সাথে তাল মিলিয়ে না চললে performance কমে যায় — এটা unnoticed থেকে গেলে বড় সমস্যা হতে পারে।

### Monitor না করলে কী সমস্যা হতে পারে?
Model silently খারাপ prediction দিতে থাকবে, business-এ negative impact পড়বে, কিন্তু কেউ জানবেও না কেন output খারাপ আসছে।

### Deployed NLP system-এ কী কী monitor করবে?
+ Prediction accuracy/confidence over time
+ Response latency
+ Input data distribution (data drift)
+ Error rate/failure logs
+ User feedback

## Model Update

### Model-এর কেন update দরকার?
Data, language usage, এবং user behavior সময়ের সাথে বদলায় — পুরনো model নতুন pattern ধরতে পারে না, তাই update প্রয়োজন।

### Model Drift কী?
সময়ের সাথে real-world data-এর distribution বদলে যাওয়ার কারণে model-এর performance ধীরে ধীরে কমে যাওয়া।

### Model কখন retrain করা উচিত?
+ Performance metric significantly কমে গেলে
+ New data pattern দেখা দিলে
+ নির্দিষ্ট সময় পরপর (scheduled retraining)
+ Domain/business requirement change হলে

### নতুন model পুরনোটার চেয়ে ভালো কিনা কীভাবে বুঝব?
Same test set-এ দুটো model-এর metric (accuracy, F1-score) compare করে, অথবা A/B testing করে real-world performance যাচাই করে।

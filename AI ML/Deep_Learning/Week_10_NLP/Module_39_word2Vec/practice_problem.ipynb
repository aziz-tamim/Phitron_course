## Q1. Skip-gram (window size = 1)
Rule: center word = **Input**, surrounding word(s) = **Target**

````python
sentence = "I love machine learning today"
words = sentence.split()
window = 1

pairs = []
for i, center in enumerate(words):
    for j in range(max(0, i-window), min(len(words), i+window+1)):
        if j != i:
            pairs.append((center, words[j]))

print("Input-Target Pairs:")
for p in pairs:
    print(p)
print("\nTotal pairs:", len(pairs))
````

**Expected Output:**
```
('I', 'love')
('love', 'I')
('love', 'machine')
('machine', 'love')
('machine', 'learning')
('learning', 'machine')
('learning', 'today')
('today', 'learning')

Total pairs: 8
```

### Answer 2: Input when predicting "I"
"I" is a **target**. Check which pairs have target = "I" -> input = **love**

### Answer 3: Input when predicting "machine"
Target = "machine" -> inputs = **love**, **learning**

````python
target_word = "I"

inputs_for_I = [p[0] for p in pairs if p[1] == target_word]

print(f"Input(s) when predicting '{target_word}':", inputs_for_I)

target_word2 = "machine"
inputs_for_machine = [p[0] for p in pairs if p[1] == target_word2]
print(f"Input(s) when predicting '{target_word2}':", inputs_for_machine)
````

## Q2. Word2Vec vs One-Hot Encoding
Code দিয়ে দেখি dimensionality আর memory এর difference।

````python
import numpy as np
import sys

vocab = list(set(words))
vocab_size = len(vocab)
print("Vocabulary:", vocab)
print("Vocab size:", vocab_size)

# One-Hot Encoding
one_hot_vectors = np.eye(vocab_size)
print("\nOne-Hot vector for '" + vocab[0] + "':")
print(one_hot_vectors[0])
print("One-Hot dimension:", one_hot_vectors[0].shape)
print("Memory (bytes):", one_hot_vectors.nbytes)
````

````python
# Word2Vec style dense vector (dummy example, fixed small size)
embedding_dim = 5   # real Word2Vec e.g. 100-300, choto example er jonno 5 rakhlam
word2vec_vector = np.random.rand(embedding_dim)

print("Word2Vec-style dense vector for '" + vocab[0] + "':")
print(word2vec_vector)
print("Word2Vec dimension:", word2vec_vector.shape)
print("Memory (bytes):", word2vec_vector.nbytes)

print("\n--- Comparison ---")
print(f"One-Hot dim grows with vocab size -> {vocab_size}")
print(f"Word2Vec dim stays fixed -> {embedding_dim} (even if vocab size = 100,000)")
````

**Summary Table:**

| Criteria | One-Hot | Word2Vec |
|---|---|---|
| Dimensionality | = vocab size (grows) | Fixed (e.g. 100-300) |
| Similarity | Can't capture | Captures semantic similarity |
| Memory | High (sparse) | Low (dense, compact) |

## Q3. CBOW vs Skip-gram (using gensim library)
Real Word2Vec model দিয়ে দেখানো হলো — dুইটা mode কিভাবে train হয়।

````python
from gensim.models import Word2Vec

sentences = [words]   # gensim needs list of tokenized sentences

# sg=0 -> CBOW, sg=1 -> Skip-gram
cbow_model = Word2Vec(sentences, vector_size=10, window=1, min_count=1, sg=0)
skipgram_model = Word2Vec(sentences, vector_size=10, window=1, min_count=1, sg=1)

print("CBOW vector for 'machine':\n", cbow_model.wv['machine'])
print("\nSkip-gram vector for 'machine':\n", skipgram_model.wv['machine'])
````

**Difference:**

| | CBOW | Skip-gram |
|---|---|---|
| Input | Context words | Target (center) word |
| Output | Target word | Context words |
| Direction | Context -> Target | Target -> Context |
| Good for | Frequent words, faster | Rare words, large dataset |

✅ **Skip-gram predicts context from target word** (`sg=1` in gensim)
❌ CBOW does the opposite (context -> target)
